## 3. Quick look at the data (2-3 real numbers)

Loading the starter dataset below to check whether there's actually enough volume in this lane to justify the next 7 weeks.

```python
import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

print("Total pages:", len(df))
print(df["trend_direction"].value_counts())

declining_with_demand = df[(df.trend_direction == "down") & (df.impressions_90d >= 100)]
print(f"\nDeclining pages with real demand (impressions_90d >= 100): {len(declining_with_demand)}")

low_ctr_visible = df[(df.impressions_90d >= 500) & (df.avg_position.between(0, 20, inclusive='right')) & (df.ctr < 0.5)]
print(f"Visible pages under-capturing clicks: {len(low_ctr_visible)}")
```
